```python
#Write a file using ; as the delimiter and -999 for missing values.
#Read it so -999 becomes NaN.

data = {"A": [10, -999, 30], "B": [40, 50, -999]}
df_out = pd.DataFrame(data)

file_path = "data_missing.csv"
df_out.to_csv(file_path, sep=";", index=False)

df_in = pd.read_csv(file_path, sep=";", na_values=[-999])
print(df_in)
```

```python
#Read a big CSV in chunks of 2000 rows and compute the mean of one column per key across all chuks
chunk_size = 2000
file_path = "titanic.csv"  # Replace with your actual dataset filename if needed

# Initialize accumulators to store sums and counts across chunks
total_sum = None
total_count = None

for chunk in pd.read_csv("/content/tested.csv", chunksize=chunk_size):
  grouped = chunk.groupby("Pclass")["Fare"]

  chunk_sum = grouped.sum()
  if total_sum is None:
    total_sum = chunk_sum
  else:
    total_sum = total_sum.add(chunk_sum, fill_value=0)

  chunk_count = grouped.count()
  if total_count is None:
    total_count = chunk_count
  else:
    total_count = total_count.add(chunk_count, fill_value=0)

overall_mean = total_sum / total_count

print("Overall Mean Fare per Passenger Class (Pclass):")
print(overall_mean)
```

```python
#Export a Dataframe with orient="records",read it back with pd.read_json, and check equality with .equals

df_original = pd.DataFrame({"Name": ["Alice", "Bob"], "Score": [88, 92]})

json_str = df_original.to_json(orient="records")

df_read = pd.read_json(io.StringIO(json_str), orient="records")
are_equal = df_original.equals(df_read)

print(f"DataFrames are equal: {are_equal}")
```

```python
import json
json_text = """
{
  "shop": "Fresh Mart",
  "open": true,
  "owner": null,
  "items": [
    {"name": "Rice", "price": 60, "qty": 10},
    {"name": "Oil", "price": 150, "qty": 4},
    {"name": "Tea", "price": 120, "qty": 6}
  ]
}
"""
t = json.loads(json_text)
print("1. Type of t:", type(t))
second_item_name = t["items"][1]["name"]
num_items = len(t["items"])
print(f"2. Second item name: {second_item_name}")
print(f"   Number of items: {num_items}")

owner_value = t["owner"]
print(f"3. Value of owner: {owner_value}")
print(f"   Python value type: {type(owner_value)}")

json_back = json.dumps(t, indent=2)
print("4. Converted back to JSON text:\n")
print(json_back)
```

```python
import pandas as pd

items_list = t["items"]

df = pd.DataFrame(items_list)
print("1. DataFrame:")
print(df)
print("-" * 30)

df["total"] = df["price"] * df["qty"]
print("2. DataFrame with 'total' column:")
print(df)
print("-" * 30)

grand_total = df["total"].sum()
max_total_index = df["total"].idxmax()
highest_total_item = df.loc[max_total_index, "name"]

print(f"3. Grand Total: {grand_total}")
print(f"   Item with the highest total: {highest_total_item}")
```
